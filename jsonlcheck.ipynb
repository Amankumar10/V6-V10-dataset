```python
import json

file_path = "G:\V5-dataset\isha\data5a2k.jsonl"

def validate_jsonl(file_path):
    valid = 0
    invalid = 0

    with open(file_path, "r", encoding="utf-8") as f:
        for i, line in enumerate(f, start=1):
            line = line.strip()

            if not line:
                continue

            try:
                obj = json.loads(line)

                if not isinstance(obj, dict):
                    print(f"Line {i}: Not a JSON object")
                    invalid += 1
                else:
                    valid += 1

            except json.JSONDecodeError as e:
                print(f"Line {i}: {e}")
                invalid += 1

    print(f"\nValid lines: {valid}")
    print(f"Invalid lines: {invalid}")

    return invalid == 0


# Example
if validate_jsonl(file_path):
    print("✅ JSONL dataset is valid")
else:
    print("❌ JSONL dataset contains errors")
```
