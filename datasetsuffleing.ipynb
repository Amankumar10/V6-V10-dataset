```python
import json
import random

input_file = "./V5A.jsonl"
output_file = "finalV4B_shuffled.jsonl"

with open(input_file, "r", encoding="utf-8") as f:
    lines = f.readlines()

random.seed(42)
random.shuffle(lines)

with open(output_file, "w", encoding="utf-8") as f:
    f.writelines(lines)

print(f"Saved {output_file}")
```
