```python
import json
from collections import Counter

# FILE_PATH = r"G:\V5-dataset\V5C\V5C.jsonl"
FILE_PATH = r"G:\V5-dataset\V5B\V5B.jsonl"


message_counts = []

with open(FILE_PATH, "r", encoding="utf-8") as f:
    for line in f:
        if not line.strip():
            continue

        record = json.loads(line)

        conversation = record["input"]["conversation"]

        # Each line = one message
        message_count = len(conversation.split("\n"))

        message_counts.append(message_count)


# ============================================================
# BASIC VALUES
# ============================================================

total = len(message_counts)

minimum = min(message_counts)
maximum = max(message_counts)
average = sum(message_counts) / total


# ============================================================
# PERCENTAGE DISTRIBUTION
# ============================================================

counts = Counter(message_counts)

min_percentage = (counts[minimum] / total) * 100
max_percentage = (counts[maximum] / total) * 100


# Average as % of maximum
average_percentage_of_max = (average / maximum) * 100


# ============================================================
# OUTPUT
# ============================================================

print("=" * 70)
print("CONVERSATION LENGTH — PERCENTAGE ANALYSIS")
print("=" * 70)

print(f"Total conversations : {total}")

print()
print(f"MINIMUM")
print(f"Messages            : {minimum}")
print(f"Percentage          : {min_percentage:.2f}%")

print()
print(f"AVERAGE")
print(f"Messages            : {average:.2f}")
print(f"% of maximum        : {average_percentage_of_max:.2f}%")

print()
print(f"MAXIMUM")
print(f"Messages            : {maximum}")
print(f"Percentage          : {max_percentage:.2f}%")

print("=" * 70)

print("\nFULL PERCENTAGE DISTRIBUTION")
print("=" * 70)

for length in sorted(counts):
    count = counts[length]
    percentage = (count / total) * 100

    print(
        f"{length:>3} messages : "
        f"{count:>8} conversations : "
        f"{percentage:>6.2f}%"
    )

print("=" * 70)
```

```python
import json
from collections import Counter

FILE_PATH = r"G:\V5-dataset\V5C-Bittu.jsonl"

message_counts = []

with open(FILE_PATH, "r", encoding="utf-8") as f:
    for line in f:
        if not line.strip():
            continue

        record = json.loads(line)

        conversation = record["input"]["conversation"]

        # Count messages
        messages = conversation.split("\n")
        message_count = len(messages)

        message_counts.append(message_count)


# ============================================================
# CALCULATIONS
# ============================================================

total = len(message_counts)

minimum = min(message_counts)
maximum = max(message_counts)
average = sum(message_counts) / total

counts = Counter(message_counts)


# ============================================================
# OUTPUT
# ============================================================

print("=" * 75)
print("CONVERSATION LENGTH — PERCENTAGE ANALYSIS")
print("=" * 75)

print(f"Total conversations : {total}")

print()
print("MINIMUM")
print(f"Messages            : {minimum}")
print(f"Conversations       : {counts[minimum]}")
print(f"Percentage          : {(counts[minimum] / total) * 100:.4f}%")

print()
print("AVERAGE")
print(f"Messages            : {average:.2f}")
print(f"% of maximum        : {(average / maximum) * 100:.2f}%")

print()
print("MAXIMUM")
print(f"Messages            : {maximum}")
print(f"Conversations       : {counts[maximum]}")
print(f"Percentage          : {(counts[maximum] / total) * 100:.4f}%")

print("=" * 75)

print("\nFULL PERCENTAGE DISTRIBUTION")
print("=" * 75)

percentage_total = 0

for length in sorted(counts):

    count = counts[length]
    percentage = (count / total) * 100

    percentage_total += percentage

    print(
        f"{length:>3} messages : "
        f"{count:>8} conversations : "
        f"{percentage:>8.4f}%"
    )

print("=" * 75)
print(f"TOTAL PERCENTAGE     : {percentage_total:.4f}%")
print("=" * 75)
```
