```python
import json

input_file = "V5A.jsonl"
output_file = "V5A_clean.jsonl"

removed = 0
kept = 0
removed_lines = []

with open(input_file, "r", encoding="utf-8") as infile, \
     open(output_file, "w", encoding="utf-8") as outfile:

    for line_number, line in enumerate(infile, start=1):
        line = line.strip()

        if not line:
            continue

        try:
            sample = json.loads(line)
        except json.JSONDecodeError:
            removed += 1
            removed_lines.append((line_number, "Invalid JSON"))
            continue

        # Remove samples with missing or empty output
        if "output" not in sample or not sample["output"]:
            removed += 1
            removed_lines.append((line_number, "Missing/empty output"))
            continue

        outfile.write(json.dumps(sample, ensure_ascii=False) + "\n")
        kept += 1

print("Done.")
print(f"Kept samples: {kept}")
print(f"Removed samples: {removed}")

if removed_lines:
    print("\nRemoved lines:")
    for line_number, reason in removed_lines:
        print(f"Line {line_number}: {reason}")

print(f"\nClean dataset saved as: {output_file}")
```
